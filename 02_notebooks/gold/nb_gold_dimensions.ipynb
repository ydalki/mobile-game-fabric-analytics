```python
player_metrics = spark.table("lh_gold_game.player_metrics")

dim_channel = (
    player_metrics
    .select("acquisition_channel")
    .distinct()
    .orderBy("acquisition_channel")
)

display(dim_channel)
```

```python
dim_channel.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("lh_gold_game.dim_channel")
```

```python
from pyspark.sql import functions as F

daily_kpis = spark.table("lh_gold_game.daily_kpis")
retention_cohorts = spark.table("lh_gold_game.retention_cohorts")

all_dates = (
    daily_kpis
    .select(F.col("activity_date").alias("date"))
    .union(
        retention_cohorts
        .select(F.col("install_date").alias("date"))
    )
    .distinct()
)

dim_date = (
    all_dates
    .withColumn("year", F.year("date"))
    .withColumn("month_number", F.month("date"))
    .withColumn("month_name", F.date_format("date", "MMMM"))
    .withColumn("week_of_year", F.weekofyear("date"))
    .withColumn("day_of_month", F.dayofmonth("date"))
    .withColumn("day_name", F.date_format("date", "EEEE"))
    .orderBy("date")
)

display(dim_date)
```

```python
dim_date.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("lh_gold_game.dim_date")
```
